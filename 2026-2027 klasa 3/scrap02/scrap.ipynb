```python
import requests as r, time, json

API = "https://quotes.toscrape.com/api/quotes"

quotes, page = [], 1
while True:
    r = requests.get(API, params={"page": page}, timeout=20)
    # time.sleep(0.5)
    r.raise_for_status()
    data = r.json()
    quotes.extend(data['quotes'])
    if not data['has_next']:
        break
    page+=1


print(len(quotes))
# print(json.dumps(quotes, indent=4, ensure_ascii=False))
with open("quotes.json", "w", encoding="utf-8") as plik:
    json.dump(quotes, plik, ensure_ascii=False, indent=4)
```
